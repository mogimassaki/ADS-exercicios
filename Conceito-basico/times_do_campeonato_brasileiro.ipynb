```python
#Entradas
dados = input('Digite o nome do time: ').lower()
#Processamento
normalizado = dados.title()
tem_fc = 'fc' in dados
comeca_s = dados[0] == 's'
termina_ense = dados.endswith('ense')
tem_clube = 'clube' in dados
tem_atletico = 'atletico' in dados
nome_curto = len(dados.strip()) < 10
nome_valido = dados.replace(' ', '').isalnum()
tem_gremio = 'grêmio' in dados or 'gremio' in dados
destaque = (tem_fc or tem_clube or tem_gremio) and not nome_curto
#Saidas
print(f'NOME_NORMALIZADO: {normalizado}')
print(f'TEM_FC: {tem_fc}')
print(f'COMECA_COM_S: {comeca_s}')
print(f'TERMINA_COM_ENSE: {termina_ense}')
print(f'TEM_CLUBE: {tem_clube}')
print(f'TEM_ATLETICO: {tem_atletico}')
print(f'NOME_CURTO: {nome_curto}')
print(f'NOME_VALIDO: {nome_valido}')
print(f'DESTAQUE: {destaque}')
```
