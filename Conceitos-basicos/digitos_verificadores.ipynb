```python
dados_entrada = 'Tinga Bank;1020;654321-00'.split(';') # Correção: Dividir por ponto e vírgula
#dados_entrada = input().split(';')
banco = dados_entrada[0].upper()
agencia = dados_entrada[1]
conta_completa_str = dados_entrada[2] # Usar uma nova variável para a string completa da conta '654321-00'

numero_conta_str = conta_completa_str.split('-')[0] # Será '654321'

# Processamento para DV1
soma_ponderada_dv1 = ((int(numero_conta_str[0]) * 7) + (int(numero_conta_str[1]) * 6) + (int(numero_conta_str[2]) * 5) + (int(numero_conta_str[3]) * 4) + (int(numero_conta_str[4]) * 3) + (int(numero_conta_str[5]) * 2))
mod11 = soma_ponderada_dv1 % 11
# Aqui usamos valores booleanos para definir o valor de dv1
mod11_booleano = mod11 != 10
dv1_calculado = mod11 * mod11_booleano # Armazenar o dígito calculado em uma nova variável

soma_ponderada_dv2 = ((int(numero_conta_str[0]) * 8) + (int(numero_conta_str[1]) * 7) + (int(numero_conta_str[2]) * 6)
+ (int(numero_conta_str[3]) * 5) + (int(numero_conta_str[4]) * 4) + (int(numero_conta_str[5]) * 3) + (dv1_calculado) * 2) # Usar dv1_calculado aqui
mod11_dv2 = soma_ponderada_dv2 % 11
mod11_booleano_dv2 = mod11_dv2 != 10
dv2_calculado = mod11_dv2 * mod11_booleano_dv2 # Armazenar o dígito calculado em uma nova variável

dv_esperado = str(dv1_calculado) + str(dv2_calculado)
dv_informado = conta_completa_str.split('-')[1]
dv_ok = dv_esperado == dv_informado
conta_valida = dv_ok
# Saídas
print(f'RESUMO: BANCO {banco} — AGENCIA {agencia}')
print(f'CONTA: {conta_completa_str}')
print(f'PESOS_D1: 7,6,5,4,3,2 SOMA={soma_ponderada_dv1} MOD11={mod11} D1={dv1_calculado}')
print(f'PESOS_D2: 8,7,6,5,4,3,2(D1) SOMA={soma_ponderada_dv2} MOD11={mod11_dv2} D2={dv2_calculado}')
print(f'DV_ESPERADO: {dv_esperado}')
print(f'DV_INFORMADO: {dv_informado}')
print(f'DV_OK: {dv_ok}')
print(f'CONTA_VALIDA: {conta_valida}')
```
