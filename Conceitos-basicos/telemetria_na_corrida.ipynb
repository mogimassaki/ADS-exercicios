```python
# #dados = input()
dados = 'R-77;DevRacing;102.3;215;6.0;52;120;120;50;35;5'.split(';')
# carro = dados[0]
# equipe = dados[1]
# tempo_volta_s = float(dados[2])
# vel_media_kmh = float(dados[3])
# dist_volta_km = float(dados[4])
# voltas_gp = int(dados[5])
# tanque_cap_l = float(dados[6])
# combustivel_atual_l = float(dados[7])
# consumo_100km = float(dados[8])
# abast_por_pit_l = float(dados[9])
# reserva_pct = (dados[10])
# #Processamento
# distancia_gp_km = dist_volta_km * voltas_gp
# consumo_por_volta_l = (consumo_100km / 100) * dist_volta_km
# combustivel_necessario_l = (consumo_por_volta_l * voltas_gp) + ((consumo_por_volta_l * voltas_gp) * 0.5)
# pits_estimados = (combustivel_necessario_l - combustivel_atual_l) / abast_por_pit_l
# defit = combustivel_necessario_l - combustivel_atual_l
# voltas_possivel = combustivel_atual_l / consumo_por_volta_l
# speed_coerente_5pct = '5' == reserva_pct
# #Saidas
# print(f'RESUMO: {carro} - EQUIPE: {equipe}')
# print(f'LAP_TIME_S: {tempo_volta_s:.3f}')
# print(f'VEL_MEDIA_KMH: {vel_media_kmh:.3f}')
# print(f'DISTANCIA_GP_KM: {distancia_gp_km:.3f}')
# print(f'CONSUMO_POR_VOLTA_L: {consumo_por_volta_l:.3f}')
# print(f'COMBUSTIVEL_NECESSARIO_L: {combustivel_necessario_l:.3f}')
# print(f'DEFIT_L: {defit:.3f}')
# print(f'PIT_ESTIMADOS: {int(pits_estimados)}')
# print(f'VOLTAS_POSSIVEIS_TANQUE_ATUAL: {int(voltas_possivel)}')
# print(f'SPEED_COERENTE_5PCT: {speed_coerente_5pct}')
###########################################################
# dados = input().split(';')
# #dados = 'R-77;DevRacing;102.3;215;6.0;52;120;120;50;35;5'.split(';')
# carro = dados[0]
# equipe = dados[1]
# tempo_volta_s = float(dados[2])
# vel_media_kmh = float(dados[3])
# dist_volta_km = float(dados[4])
# voltas_gp = int(dados[5])
# tanque_cap_l = float(dados[6])
# combustivel_atual_l = float(dados[7])
# consumo_100km = float(dados[8])
# abast_por_pit_l = float(dados[9])
# reserva_pct = (dados[10])
# #Processamento
# distancia_gp_km = dist_volta_km * voltas_gp
# consumo_por_volta_l = (consumo_100km / 100) * dist_volta_km
# combustivel_necessario_l = (consumo_por_volta_l * voltas_gp) + ((consumo_por_volta_l * voltas_gp) * 0.5)
# pits_estimados = (combustivel_necessario_l - combustivel_atual_l) / abast_por_pit_l
# defit = combustivel_necessario_l - combustivel_atual_l
# voltas_possivel = combustivel_atual_l / consumo_por_volta_l
# speed_coerente_5pct = '5' == reserva_pct
# #Saidas
# print(f'RESUMO: {carro} - EQUIPE: {equipe}')
# print(f'LAP_TIME_S: {tempo_volta_s:.3f}')
# print(f'VEL_MEDIA_KMH: {vel_media_kmh:.3f}')
# print(f'DISTANCIA_GP_KM: {distancia_gp_km:.3f}')
# print(f'CONSUMO_POR_VOLTA_L: {consumo_por_volta_l:.3f}')
# print(f'COMBUSTIVEL_NECESSARIO_L: {combustivel_necessario_l:.3f}')
# print(f'DEFIT_L: {defit:.3f}')
# print(f'PIT_ESTIMADOS: {int(pits_estimados)}')
# print(f'VOLTAS_POSSIVEIS_TANQUE_ATUAL: {int(voltas_possivel)}')
# print(f'SPEED_COERENTE_5PCT: {speed_coerente_5pct}')

import math

# Entrada de dados
#dados = input().split(';')

carro = dados[0]
equipe = dados[1].upper() # Garante letras maiúsculas
tempo_volta_s = float(dados[2])
vel_media_kmh = float(dados[3])
dist_volta_km = float(dados[4])
voltas_gp = int(dados[5])
tanque_cap_l = float(dados[6])
combustivel_atual_l = float(dados[7])
consumo_100km = float(dados[8])
abast_por_pit_l = float(dados[9])
reserva_pct = float(dados[10]) # Convertido para float

# Processamento
distancia_gp_km = dist_volta_km * voltas_gp
consumo_por_volta_l = (consumo_100km / 100) * dist_volta_km

# Combustível necessário considerando a margem de segurança do enunciado
combustivel_necessario_l = (consumo_por_volta_l * voltas_gp) * (1 + reserva_pct / 100)

# O déficit não pode ser negativo (mínimo zero)
deficit_l = max(0.0, combustivel_necessario_l - combustivel_atual_l)

# O número de abastecimentos é o teto do déficit pelo volume por pit
pits_estimados = math.ceil(deficit_l / abast_por_pit_l)

# Quantidade de voltas possíveis (apenas a parte inteira)
voltas_possiveis = int(combustivel_atual_l / consumo_por_volta_l)

# Cálculo da velocidade real (distância / tempo em horas) e validação da tolerância de 5%
vel_calculada = dist_volta_km / (tempo_volta_s / 3600)
speed_coerente_5pct = abs(vel_calculada - vel_media_kmh) <= (vel_media_kmh * 0.05)

# Saídas estritamente formatadas
print(f'RESUMO: {carro} — EQUIPE: {equipe}')
print(f'LAP_TIME_S: {tempo_volta_s:.3f}')
print(f'VEL_MEDIA_KMH: {vel_media_kmh:.3f}')
print(f'DISTANCIA_GP_KM: {distancia_gp_km:.3f}')
print(f'CONSUMO_POR_VOLTA_L: {consumo_por_volta_l:.3f}')
print(f'COMBUSTIVEL_NECESSARIO_L: {combustivel_necessario_l:.3f}')
print(f'DEFICIT_L: {deficit_l:.3f}')
print(f'PITS_ESTIMADOS: {pits_estimados}')
print(f'VOLTA_POSSIVEIS_TANQUE_ATUAL: {voltas_possiveis}')
print(f'SPEED_COERENTE_5PCT: {speed_coerente_5pct}')
```
