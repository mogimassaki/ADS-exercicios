```python
# Entrada (Lê como float para aceitar números reais)
dados = float(input())

# --- PROCESSAMENTO ---

# 1. Escala de bits (Decimal - SI: potências de 10)
b = dados
kb = dados / 10**3
mb = dados / 10**6
gb = dados / 10**9
tb = dados / 10**12

# 2. Escala de bytes (Decimal - SI: potências de 10, lembrando que 1 B = 8 b)
B = dados / 8
kB = B / 10**3
MB = B / 10**6
GB = B / 10**9
TB = B / 10**12

# 3. Escala binária (IEC: potências de 1024)
KiB = B / 1024
MiB = B / 1024**2
GiB = B / 1024**3
TiB = B / 1024**4

# --- SAÍDA ---
# Função auxiliar para formatar com até 6 casas decimais sem zeros extras no final
def formatar(valor):
    return f"{round(valor, 6):.6f}".rstrip('0').rstrip('.')

print(f"ENTRADA: {formatar(b)} b")
print(f"b: {formatar(b)}")
print(f"kb: {formatar(kb)}")
print(f"mb: {formatar(mb)}")
print(f"gb: {formatar(gb)}")
print(f"tb: {formatar(tb)}")
print(f"B: {formatar(B)}")
print(f"kB: {formatar(kB)}")
print(f"MB: {formatar(MB)}")
print(f"GB: {formatar(GB)}")
print(f"TB: {formatar(TB)}")
print(f"KiB: {formatar(KiB)}")
print(f"MiB: {formatar(MiB)}")
print(f"GiB: {formatar(GiB)}")
print(f"TiB: {formatar(TiB)}")
```
