```python
#Entradas
#dados = 'bob;bob-1234-BOB-!@#'.split(';')
dados = input().split(';')
usuario = dados[0]
senha = dados[1]
#Processamento
tamanho = len(senha)
comprimento = 12 <= len(senha) <= 64
maiuscula = senha.lower() != senha
minuscula = senha.upper() != senha
digito = digito = (
    ('0' in senha) or ('1' in senha) or ('2' in senha) or ('3' in senha) or
    ('4' in senha) or ('5' in senha) or ('6' in senha) or ('7' in senha) or
    ('8' in senha) or ('9' in senha)
)
especial = ('!' in senha) or ('@' in senha) or ('#' in senha) or ('$' in senha) or ('%' in senha) or ('^' in senha) or ('&' in senha) or ('*' in senha) or ('(' in senha) or (')' in senha) or ('-' in senha) or ('_' in senha) or ('+' in senha) or ('=' in senha) or ('.' in senha) or ('?' in senha)
sem_espaco = (' ' not in senha) and ('  ' not in senha)
nao_contem_usuario = usuario.upper() not in senha.upper()
nao_contem_tinga = 'TINGA' not in senha.upper()
politica_atendida = comprimento and maiuscula and minuscula and digito and especial and sem_espaco and nao_contem_usuario and nao_contem_tinga
pontuacao = comprimento + maiuscula + minuscula + digito + especial + sem_espaco + nao_contem_usuario + nao_contem_tinga
#Saida
print(f'RESUMO: TINGA Security — Usuário: {usuario.upper()} — Tamanho: {tamanho}')
print(f'COMPRIMENTO_OK: {comprimento}')
print(f'MAIUSCULA: {maiuscula}')
print(f'MINUSCULA: {minuscula}')
print(f'DIGITO: {digito}')
print(f'ESPECIAL: {especial}')
print(f'SEM_ESPACOS: {sem_espaco}')
print(f'NAO_CONTEM_USUARIO: {nao_contem_usuario}')
print(f'NAO_CONTEM_TINGA: {nao_contem_tinga}')
print(f'POLITICA_ATENDIDA: {politica_atendida}')
print(f'PONTUACAO: {pontuacao}/8')

```
