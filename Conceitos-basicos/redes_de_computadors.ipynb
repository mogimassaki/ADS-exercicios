```python
#Entradas
# dados = '192.168.1.10 192.168.1.255 TCP 1514 64 SYN,ACK'.split()
dados = input().split()
#Processamento
resumo = dados[2] + ' ' + dados[0] + ' ' + '->' + ' ' + dados[1]
proto_normalizado = dados[2].upper()
mesma_subrede = (dados[0].split('.'))[0] == (dados[1].split('.'))[0] and (dados[0].split('.'))[1] == (dados[1].split('.'))[1] and (dados[0].split('.'))[2] == (dados[1].split('.'))[2]
broadcast = (dados[1].split('.'))[3] == '255'
src_privado = (dados[0].split('.')[0] == '10' or dados[0].startswith('192.168.') or (dados[0].split('.')[0] == '172' and 16 <= int(dados[0].split('.')[1]) <= 31))
pacote_grande = int(dados[3]) > 1500
ttl_baixo = int(dados[4]) <= 10
tpc_controle = ('SYN' in (dados[5]).upper() or 'ACK' in (dados[5]).upper()) and proto_normalizado == 'TCP'
#Saidas
print(f'RESUMO: {resumo}')
print(f'PROTO_NORMALIZADO: {proto_normalizado}')
print(f'MESMA_SUBREDE_24: {mesma_subrede}')
print(f'BROADCAST_DST_24: {broadcast}')
print(f'SRC_PRIVADA: {src_privado}')
print(f'PACOTE_GRANDE: {pacote_grande}')
print(f'TTL_BAIXO: {ttl_baixo}')
print(f'TCP_CONTROLE: {tpc_controle}')
```
