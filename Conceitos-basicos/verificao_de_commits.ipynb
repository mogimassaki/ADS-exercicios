```python
# #Entrada
# #dados = input('Digite: ')
# dados = 'docs;readme;atualiza instruções de build;0.1.0'.split(';')
# #Processamento
# resumo = f'{(dados[0]).upper()} {dados[1]} - {dados[2]}'
# tipo_valido = dados[0] in ['feat', 'fix', 'docs', 'chore', 'test', 'refactor']
# escopo = dados[0] != '' and ' ' not in dados[0].split() and dados[0].isalnum()
# mensagem = 10 <= len(dados[2]) <= 72
# versao = dados[3].count('.') == 2 and (dados[3].split('.')[0].isdigit() and dados[3].split('.')[1].isdigit() and dados[3].split('.')[2].isdigit())
# tem_issue = '#' in dados[2] and (dados[2][dados[2].find('#') + 1]).isdigit()
# breaking = '!' in dados[0] and dados[2].islower()
# #Saidas
# print(f'RESUMO: {resumo}')
# print(f'TIPO_VALIDO: {tipo_valido}')
# print(f'ESCOPO_VALIDO: {escopo}')
# print(f'MSG_TAMANHO_OK: {mensagem}')
# print(f'SEMVER_FORMATO_OK: {versao}')
# print(f'TEM_ISSUE_ID: {tem_issue}')
# print(f'BREAKING: {breaking}')

# #Entrada
# dados = input().split(';')
# #dados = 'docs;readme;atualiza instruções de build;0.1.0'.split(';')
# #Processamento
# resumo = f'{(dados[0]).upper()} {dados[1]} - {dados[2]}'
# tipo_valido = dados[0] in ['feat', 'fix', 'docs', 'chore', 'test', 'refactor']
# escopo = dados[1] != '' and ' ' not in dados[1].split() and dados[1].isalnum()
# mensagem = 10 <= len(dados[2]) <= 72
# versao = dados[3].count('.') == 2 and (dados[3].split('.')[0].isdigit() and dados[3].split('.')[1].isdigit() and dados[3].split('.')[2].isdigit())
# tem_issue = '#' in dados[2] and (dados[2][dados[2].find('#') + 1]).isdigit()
# breaking = '!' in dados[0] and dados[2].islower()
# #Saidas
# print(f'RESUMO: {resumo}')
# print(f'TIPO_VALIDO: {tipo_valido}')
# print(f'ESCOPO_VALIDO: {escopo}')
# print(f'MSG_TAMANHO_OK: {mensagem}')
# print(f'SEMVER_FORMATO_OK: {versao}')
# print(f'TEM_ISSUE_ID: {tem_issue}')
# print(f'BREAKING: {breaking}')


#Soloução dada por IA. Tenho que dar uma analisada mais tarde
# Entrada
dados = input().split(';')

# --- Processamento ---

# Separando as variáveis para o código ficar limpo e legível
tipo_bruto = dados[0]
escopo_texto = dados[1]
msg_texto = dados[2]
versao_texto = dados[3]

# 1. Resumo (Mantém o tipo com a exclamação se houver, conforme padrão convencional)
resumo = f"{tipo_bruto.upper()} {escopo_texto} - {msg_texto}"

# 2. Tipo Válido (Remove o '!' temporariamente apenas para checar se a palavra é permitida)
tipo_limpo = tipo_bruto.replace('!', '')
tipo_valido = tipo_limpo in ['feat', 'fix', 'docs', 'chore', 'test', 'refactor']

# 3. Escopo Válido (Não vazio, sem espaços, aceita letras, números, '-' e '_')
# Criamos uma função rápida ou limpamos os caracteres permitidos para testar o resto
escopo_limpo = escopo_texto.replace('-', '').replace('_', '')
escopo_valido = escopo_texto != '' and ' ' not in escopo_texto and (escopo_limpo.isalnum() or escopo_limpo == '')

# 4. Tamanho da Mensagem
mensagem_ok = 10 <= len(msg_texto) <= 72

# 5. Formato SemVer OK (Exatamente 2 pontos e as 3 partes estritamente numéricas)
partes_versao = versao_texto.split('.')
versao_ok = (versao_texto.count('.') == 2 and
             len(partes_versao) == 3 and
             partes_versao[0].isdigit() and
             partes_versao[1].isdigit() and
             partes_versao[2].isdigit())

# 6. Tem ID de Issue (Verificação com fatiamento seguro para evitar IndexError se '#' estiver no final)
pos_hashtag = msg_texto.find('#')
tem_issue = '#' in msg_texto and msg_texto[pos_hashtag + 1 : pos_hashtag + 2].isdigit()

# 7. Breaking Change (Se o tipo contém '!' OU a mensagem contém "BREAKING CHANGE")
breaking = '!' in tipo_bruto or 'BREAKING CHANGE' in msg_texto

# --- Saídas ---
print(f'RESUMO: {resumo}')
print(f'TIPO_VALIDO: {tipo_valido}')
print(f'ESCOPO_VALIDO: {escopo_valido}')
print(f'MSG_TAMANHO_OK: {mensagem_ok}')
print(f'SEMVER_FORMATO_OK: {versao_ok}')
print(f'TEM_ISSUE_ID: {tem_issue}')
print(f'BREAKING: {breaking}')
```
