```python
#Entradas
identificador = input()
leitura_anterior = int(input())
leitura_atual = int(input())
tarifa = float(input())
custo_distribuicao = float(input())
iluminacao = float(input())

#Processamento
consumo = leitura_atual - leitura_anterior
valor_energia = consumo * tarifa
valor_distribuicao = consumo * custo_distribuicao
total_conta = valor_energia + valor_distribuicao + iluminacao

#Saidas
print(f'CLIENTE {identificador}: CONSUMO {consumo} KWH')
print(f'ENERGIA: R$ {valor_energia:.2f}')
print(f'DISTRIBUICAO: R$ {valor_distribuicao:.2f}')
print(f'ILUMINACAO: R$ {iluminacao:.2f}')
print(f'TOTAL: R$ {total_conta:.2f}')
```
