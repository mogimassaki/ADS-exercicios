```python
#Entradas
segundos = int(input())
temperatura = float(input())

#Processamento
horas = segundos // 3600
segundos = segundos % 3600

minutos = segundos // 60
segundos = segundos % 60

#Saida
print(f'Horário: {horas}:{minutos}:{segundos}')
print(f'Temperatura: {temperatura:.1f} graus')
```
