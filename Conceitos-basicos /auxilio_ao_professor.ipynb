```python
#Entradas
nome = input('nome: ').upper()
id = int(input('id: '))
pontuacao_max = int(input('Pontuação maxima: '))
pontuacao_obtida = float(input('Pontuacao obtida: '))
nota_corte = float(input('Nota de corte: '))
#Processamento
percentual = (pontuacao_obtida / pontuacao_max) * 100
atingiu_corte = nota_corte <= percentual
#Saidas
print(f'ALUNO: {nome}')
print(f'QUESTAO: Q{id:03}')
print(f'RESULTADO: {pontuacao_obtida}/{pontuacao_max} pontos')
print(f'PERCENTUAL: {percentual:.2f}%')
print(f'ATINGIU_CORTE: {atingiu_corte}')
```
