```python
'''
Uma fintech deseja simular uma ação promocional para clientes
de bancos existentes no Brasil. O programa deve ler o nome do
banco e o valor de uma transferência recebida e calcular o
crédito promocional que será devolvido ao cliente. O crédito
depende apenas do banco informado.
A promoção envolve osseguintes parâmetros, se o banco for
BANCO DO BRASIL, o crédito é de 1% do valor recebido. Se for
CAIXA, o crédito é de 0,6%. Se for ITAU, o crédito é de 2%.
Se for NUBANK, o crédito é de 1,5%. Se for BRADESCO, o crédito
é de 0,75%. Em qualquer outro banco, o credito é de 1,25%. O
valor final creditado ao cliente é a soma do valor recebido
com o credito promocional.
Não existe entradas ou dados inválidos. o nome do banco pode
ser digitado em maiúscula ou minúscula. O valor da tranferência
pode usar ponto ou virgula como separador decimal.
ENTRADA
A entrada possui duas linhas. A primeira linha contém o nome
do banco. A segunda linha contém o valor da tranferência
recebida em reais (duas casas decimais). Observe que o valor
pode utilizar vírgula ou ponto como separador decimal.
SAIDA
A saida deve conter exatamente quatro linhas, contendo as
informações: BANCO <nome_formatado>; VALOR TRANSFERÊNCIA R$
 <valor_transferencia>; CRÉDITO R$ <valor_credito>; TOTAL R$
  <valor_final>. O nome do banco deve ser exibido em maiúsculas.
  Os valores numéricos devem ser mostrados com duas casas decimais.
'''
#Entrada
nome_formatado = input().upper()
valor_transferencia = float(input().replace(',', '.'))
#Processamento
if nome_formatado == 'BANCO DO BRASIL':
  valor_credito = valor_transferencia * .01
elif nome_formatado == 'CAIXA':
  valor_credito = valor_transferencia * .006
elif nome_formatado == 'ITAU':
  valor_credito = valor_transferencia * .02
elif nome_formatado == 'NUBANK':
  valor_credito = valor_transferencia * .015
elif nome_formatado == 'BRADESCO':
  valor_credito = valor_transferencia * 0.0075
else:
  valor_credito = valor_transferencia * .0125
valor_final = valor_transferencia + valor_credito
#Saida
print(f'BANCO {nome_formatado}')
print(f'VALOR TRANSFERÊNCIA R$ {valor_transferencia:.2f}')
print(f'CRÉDITO R$ {valor_credito:.2f}')
print(f'TOTAL R$ {valor_final:.2f}')
```
