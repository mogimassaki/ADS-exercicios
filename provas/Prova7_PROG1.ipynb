```python
'''
Você precisa criar um programa em Python 3 para permitir
que um usúario cadastre uma nova senha em um sistema de
segurança. O sistema deve aceitar até 5 tentativas, uma
por linha, e verificar se alguma delas atende a todos os
critérios exigidos. A senha será aceita somente quando
possuir: (i) pelo menos 8 caracteres; (ii) pelo menos um
número; (iii) pelo menos uma letra maiúscula; e, (iv)
pelo menos uma letra minúscula.
Quando uma tentativa não atender os critérios, o programa
deve informar o que faltou, sempre nesta ordem: tamanho
minimo, número, letra maiúscula e letra minúscula. Se mais
um critério estiver ausente, todos os problemas devem
aparecer na mesma linha, separados por ponto e vírgula.
Assim que uma senha válida for digitada, o programa deve
exibir a menssagem "Senha alterada com sucesso!" e encerrar.
Caso as 5 tentativas sejam utilizadas sem que nenhuma senha
válida seja informada, o programa deve exibir "Limite de
tentativas atingido. Conta bloqueada." (observe que deve
conter o ponto final).
ENTRADA
A entrada contém até 5 linhas, cada uma com uma tentativa
de senha. O programa deve ler as tentativas em ordem e
parar antes da quinta tentativa caso encontre uma senha
válida. Não há dados inválidos de entrada, pois cada linha
deve ser tratada apenas como texto. Além disso, não haverá
mais que 5 linhas e os casos com uma quantidade menor de
tentativas encerram com sucesso.
SAIDA
Para cada tentativa inválida, o programa deve imprimir
uma linha no formato Tentativa <n>: <menssagens>, informando
os critérios que faltam na senha. As menssagens possíveis
são: (i) "falta pelo menos 8 caracteres"; (ii) "falta um
número"; (iii) "falta uma letra maiúscula"; e, (iv) "falta
uma letra minúscula", sempre nessa ordem. Quando a senha
for aceita, deve imprimir "Senha alterada com sucesso!".
Se nenhuma das 5 tentativas for aceita, deve imprimir ao
final "Limite de tentativas atingido. Conta bloqueada.".
'''
contador = 0
senha_valida = False
while contador < 5 and senha_valida == False:
  criterio1 = criterio2 = criterio3 = criterio4 = criterio5 = False
  motivo = []
  senha = input()
#Validador de comprimento
  if len(senha) >= 8:
    criterio1 = True
  else:
    motivo.append('falta pelo menos 8 caracteres')
#Validador de numeros
  resultados = []
  for n in senha:
    resultados.append(n.isdigit())
  if any(resultados):
    criterio2 = True
  else:
    motivo.append('falta um número')
#Validador de Maiusculas
  if senha != senha.lower():
    criterio3 = True
  else:
    motivo.append('falta uma letra maiúscula')
#Validador de Minuscula
  if senha != senha.upper():
    criterio4 = True
  else:
    motivo.append('falta uma letra minúscula')
#Validador de Senha
  if criterio1 == True and criterio2 == True and criterio3 == True and criterio4 == True:
    senha_valida = True
  else:
    contador = contador + 1
    print(f'Tentativa {contador}: {'; '.join(motivo)}')

if senha_valida == True:
  print('Senha alterada com sucesso!')
else:
  print('Limite de tentativas atingido. Conta bloqueada.')



```
