```python
'''
A organização da Copa do Mundo deseja classificar o desempenho
de uma seleção em uma partida. O programa deve receber, em uma
única linha, o nome da seleção, a fase da competição, a
quantidade de gols marcados e a quantidade de gols sofridos.
A seleção deve ser exibida em letras maiúsculas e com underlines
substituidos por espaços.
As fases válidas são GRUPOS, 16_AVOS, OITAVAS, QUARTAS, SEMIFINAL
 e FINAL. A quantidade de gols marcados e sofridos deve ser maior
 ou igual a zero. Se a fase não estiver entre as opções válidas
 ou se algum valor de gols for negativo, o programa deve imprimir
 apenas DADOS INVALIDOS (sem acento).
 Caso os dados forem válidos, o programa deve determinar o
 resultado da partida. Se os gols marcados forem maiores que os
 gols sofridos, o resultado é VITORIA. Se forem iguais, o
 resultado é EMPATE. Se forem menores o resultado é DERROTA.
 Além disso, o programa também deve indicar a situação da seleção
 conforme a fase. Na fase GRUPOS, vitória gera GANHOU_3_PONTOS,
 empate  gera GANHOU_1_PONTO e derrota SEM_PONTOS. Nas fases
 eliminatórias (16_AVOS, OITAVAS, QUARTAS, SEMIFINAL e FINAL),
 vitória gera AVANCA, empate gera DECISAO_NOS PENALTIS e derrota
 gera ELIMINADA. Na fase FINAL, se houver vitória, a situação
 deve ser CAMPEA; se houver empate, DECISAO_NOS_PENALTIS; se
 houver derrota, VICE.
 ENTRADA
 A entrada possui uma única linha com quatro valores separados
 por espaço correspondendo a um único jogo, nesta ordem:
 selecao fase gols_marcados gols_softridos. A seleção pode
 conter underlines no lugar de espaços. Os gols marcados e
 sofridos devem ser tratados como numeros inteiros, os demais
 dados como strings.
 SAIDA
 A saída contém dois formatos diferentes. Caso houver dados
 inválidos, a saída deve conter três linhas, contendo as
 informações: SELECAO=<selecao_formatada>, RESULTADO=<resultado>,
 SITUACAO=<situacao>, conforme exemplo abaixo.
 '''

lista = ['GRUPOS', '16_AVOS', 'OITAVAS', 'QUARTAS', 'SEMIFINAL', 'FINAL']
eliminatorias = ['16_AVOS', 'OITAVAS', 'QUARTAS', 'SEMIFINAL']
try:
#Entrada de Dados
  dados = input().split()
  selecao = dados[0].upper().replace('_', ' ')
  fase = dados[1].upper()
  gols_marcados = int(dados[2])
  gols_sofridos = int(dados[3])
#Processamento de Erros
  if fase not in lista:
    raise ValueError('DADOS INVALIDOS')

  if gols_marcados < 0 or gols_sofridos < 0:
    raise ValueError('DADOS INVALIDOS')

#Processamento de Resultado
  if gols_marcados < gols_sofridos:
    resultado = 'DERROTA'
  elif gols_marcados == gols_sofridos:
    resultado = 'EMPATE'
  elif gols_marcados > gols_sofridos:
    resultado = 'VITORIA'
#Processamento de Situação conforme Fase
#Fase de Grupos
  if fase == 'GRUPOS' and resultado == 'VITORIA':
    situacao = 'GANHOU_3_PONTOS'
  elif fase == 'GRUPOS' and resultado == 'EMPATE':
    situacao = 'GANHOU_1_PONTO'
  elif fase == 'GRUPOS' and resultado == 'DERROTA':
    situacao = 'SEM_PONTOS'
#Fases Eliminatoria
  elif fase in eliminatorias and resultado == 'VITORIA':
    situacao = 'AVANCA'
  elif fase in eliminatorias and resultado == 'EMPATE':
    situacao = 'DECISAO_NOS_PENALTIS'
  elif fase in eliminatorias and resultado == 'DERROTA':
    situacao = 'ELIMINADA'
#Fase Final
  elif fase == 'FINAL' and resultado == 'VITORIA':
    situacao = 'CAMPEA'
  elif fase == 'FINAL' and resultado == 'EMPATE':
    situacao = 'DECISAO_NOS_PENALTIS'
  elif fase == 'FINAL' and resultado == 'DERROTA':
    situacao = 'VICE'
  print(f'SELECAO={selecao}')
  print(f'RESULTADO={resultado}')
  print(f'SITUACAO={situacao}')
except:
  print('DADOS INVALIDOS')


```
