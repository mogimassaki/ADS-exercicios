```python
'''
A loja de dispositivos eletronicos TechCelulares deseja gerar um resumo
simples de uma venda. O seu programa em Python 3 deve ler três dados: o
nome do produto, o preço base e o selo promocional. O nome pode vir com
underlines no lugar de espaços e deve ser exibido com os espaços
restaurados (no lugar dos underlines) e em letras maiúsculas. O preço
pode ser informado com ponto ou virgula como separador decimal.
O programa deve informar o preço base, o preço final da venda e dois
indicadores logicos. O indicador MODELO_PREMIUM será verdadeiro quando
o nome do produto tiver a palavra PRO ou MAX. O indicador TEM_DESCONTO
será verdadeiro quando o selo promocional contiver a palavra VIP.
O desconto de venda será de 15% apenas quando o selo contiver a palavra
VIP. O preço final é calculado aplicado esse desconto diretamente sobre
o preço base.
ENTRADA
A entrada possui exatamente três linhas. A primeira linha contém o nome
do produto. A segunda linha contém o preço base. A  terceira linha
contém o selo promocional. O nome pode conter sublinhados no lugar de
espaços. O preço deve ser tratado como numero real  e pode usar ponto
ou virgula como separador decimal. O selo promocional é um texto e pode
conter ou não a palavra VIP.
SAÍDA
A saida deve conter exatamente cinco linhas, contendo
PRODUTO: <nome_formatado>, PREÇO BASE: <valor>, PREÇO FINAL: <valor,
MODELO PREMIUM: <True_ou_False>, TEM DESCONTO: <True_ou_False>. O
nome deve ser exibido em maiúsculas e com os underlines substituidos por
espaços. Os valorews monetários devem ser mostrados com duas casas decimais.
'''
#Entrada
nome_produto = input().replace('_', ' ').upper()
preco_base = float(input().replace(';', '.'))
selo_promocional = input().upper()
#Processamento
modelo_premium = 'PRO' in nome_produto or 'MAX' in nome_produto
tem_desconto = selo_promocional == 'VIP'
desconto = preco_base * .15 * tem_desconto
preco_final = preco_base - desconto
#Saida
print(f'PRODUTO: {nome_produto}')
print(f'PREÇO BASE: {preco_base:.2f}')
print(f'PREÇO FINAL: {preco_final:.2f}')
print(f'MODELO PREMIUM: {modelo_premium}')
print(f'TEM DESCONTO: {tem_desconto}')
```
