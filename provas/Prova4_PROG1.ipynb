```python
'''
A DEA está analisando relatórios simplificados sobre personagens
da série Breaking Bad. Seu programa deve ler informações de um
personagem e definir o perfil da operção em que ele esta inserido.
A primeira linha de entrada contém três informações separadas por
espaço: o papel pode ser um entre QUIMICO, DISTRIBUIDOR, POLICIAL
ou ADVOGADO. O dinheiro é um número real e pode ser informado
com ponto ou virgula como separador decimal. O risco pode ser
BAIXO, MEDIO, ou ALTO.
O nome deve ser exibio em maiúsculas, com os underlines substituidos
por espaços. O programa deve calcular um indicador lógico chamado
OPERACAO_GRANDE, que será verdadeiro quando o valor em dinheiro
for maior ou igual a 50000. Em seguida, deve determinar o PERFIL
da operação de acordo com o papel do personagem e o risco informado.
As regras são as seguintes: se o papel for QUIMICO, o perfil será
LABORATORIO_MAXIMO quando o risco for ALTO, LABORATORIO_CONTROLADO
quando o risco for MEDIO e LABORATORIO_LOCAL quando o risco for
BAIXO. Se o papel for DISTRIBUIDOR, o perfil será REDE_EXPANDIDA
quando a operação for grande, e REDE_LOCAL caso contrário. Se o
for POLICIAL, o perfil será INVESTIGACAO_CRITICA quando o risco
for ALTO, e INVESTIGACAO_PADRAO nos demais casos. Se o papel
for ADVOGADO, o perfil será COBERTURA_TOTAL quando a operção
for grande e o risco for ALTO, e COBERTURA_LEGAL nos demais casos.
ENTRADA
A entrada possui exatamente duas linhas. A primeira contém o nome
do personagem. A segunda linha contém três valores separados
por espaço, nesta ordem: papel, dinheiro e risco. O nome pode conter
underlines no lugar de espaços. O papel e o risco devem ser
tratados como textos. O dinheiro deve ser tratado como número
real e pode ser usado ponto e vírgula como separador decimal.
Não existem dados inválidos: todas as entradas poderão ser lidas
corretamente.
SAIDA
A saída deve conter exatamente quatro linhas, com as informações:
(i) PERSONAGEM=<nome_formatado>; (ii) DINHEIRO=<valor>;
(iii) OPERACAO_GRANDE=<True_ouFalse>; e, (iv) PERFIL=<perfil>.
O nome deve ser exibido em maiúscula e com os underlines
substituídos por espaços. O valor do dinheiro deve ser mostrado
com duas casas decimais.
'''
#Entrada
nome_formatado = input().upper().replace('_', ' ')
dados = input().split()
papel = dados[0].upper()
dinheiro = float(dados[1].replace(',', '.'))
risco = dados[2].upper()
#Processamento
if dinheiro >= 50000:
  operacao_grande = True
else:
  operacao_grande = False
if papel == 'QUIMICO' and risco == 'ALTO':
  perfil = 'LABORATORIO_MAXIMO'
elif papel == 'QUIMICO' and risco == 'MEDIO':
  perfil = 'LABORATORIO_CONTROLADO'
elif papel == 'QUIMICO' and risco == 'BAIXO':
  perfil = 'LABORATORIO_LOCAL'
elif papel == 'DISTRIBUIDOR' and operacao_grande == True:
  perfil = 'REDE_EXPANDIDA'
elif papel == 'DISTRIBUIDOR' and operacao_grande == False:
  perfil = 'REDE_LOCAL'
elif papel == 'POLICIAL' and risco == 'ALTO':
  perfil = 'INVESTIGACAO_CRITICA'
elif papel == 'POLICIAL' and (risco == 'MEDIO' or risco == 'BAIXO'):
  perfil = 'INVESTIGACAO_PADRAO'
elif papel == 'ADVOGADO' and operacao_grande == True and risco == 'ALTO':
  perfil = 'COBERTURA_TOTAL'
elif papel == 'ADVOGADO':
  perfil = 'COBERTURA_LEGAL'
#Saida
print(f'PERSONAGEM={nome_formatado}')
print(f'DINHEIRO={dinheiro:.2f}')
print(f'OPERACAO_GRANDE={operacao_grande}')
print(f'PERFIL={perfil}')
```
