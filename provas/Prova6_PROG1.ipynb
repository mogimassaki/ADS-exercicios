```python
'''
A empresa Tinga Routes utiliza dois códigos inteiros para
gerar uma sequência progressiva de controle. O programa
deve receber dois valores inteiros em uma única linha.
Quando o primeiro valor for menor que o segundo, devem ser
considerados apenas os numeros pares entre eles, incluindo
os limites quando forem pares, em ordem crescente. Quando
o primeiro valor for maior que o segundo, devem ser
considerados apenas os números impares entre eles, incluido
os limites quando forem impares, em ordem decrescente.
Quando os dois valores forem iguais, a sequência terá apenas
esse valor.
A sequência deve ser construida progressivamente: a cada
novo numero encontrado, deve ser impressa uma nova linha
mostrado todos os numeros já adicionados até aquele momento.
Não há dados inválidos.
ENTRADA
A entrada possui uma única linha com dois valores inteiros
separados por espaços, representando o codigo inicial e o
codigo final.
SAIDA
A saida deve conter uma ou mais linhas com os numeros referente
a sequencia separados por um único espaço. Cada linha deve
mostrar asequência acumulada até aquela etapa. Os valores
devem aparecer separados por um único espaço. Se os dois
valores forem iguais, imprima apenas uma linha com esse valor.
'''
dados = input().split()
num1 = int(dados[0])
num2 = int(dados[1])
lista = []
if num1 == num2:
  print(num1)
elif num1 < num2:
  for n in range(num1, num2 + 1):
    if n % 2 == 0:
      lista.append(str(n))
      print(' '.join(lista))
elif num1 > num2:
  for n in range(num1, num2 - 1, -1):
    if n % 2 != 0:
      lista.append(str(n))
      print(' '.join(lista))
```
