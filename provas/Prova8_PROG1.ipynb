```python
'''
Você precisa criar um programa em Python 3 para analisar uma
sequência de valores informado. Cada linha da entrada contém
um valor que deveria representar um número inteiro. A leitura
deve terminar quando o valor 0 for informado. O valor 0 serve
apenas para encerrar o processamento e não deve ser processado
em nenhuma categoria.
Para cada valor lido antes do 0, o programa deve verificar se
ele representa um numero inteiro. Quando o valor for inteiro,
ele deve ser contado conforme suas propiedades: positivo ou
negativo, par ou ímpar. Um mesmo número inteiro deve contribuir
para duas contagem: uma entre positivos/negativos e outra entre
par/ímpar. Quando o valor não representar um número inteiro, ele
deve ser contado como inválido e não deve participar das demais
contagens.
Ao final, o programa deve exibir a quantidade de números
positivos, negativos, pares, impares e inválidos informados
antes do encerramento.
ENTRADA
A entrada é composta por diversas linhas, cada uma contendo um
único valor. Cada valor deve ser analisado individualmente,
pois pode representar um número inteiro ou pode ser um dado
que nãopode ser convertido para inteiro. A leitura sempre será
encerrada quando for informada uma linha contendo exatamente 0.
Esse valor é apenas o sentinela de encerramento e, portanto,
não deve ser contabilizado como positivo, negativo, par, ímpar,
ou inválido.
SAIDA
A saída deve apresentar um resumo com cinco linhas, sempre na
mesma ordem. A primeira linha informa a quantidade de valores
positivos, a segunda informa a quantidade de valores negativos,
a terceira informa a quantidade de valores pares, a quarta
informa a quantidade de valores impares e a quinta informa a
quantidade de valores inválidos. Cada linha deve seguir o
formato NOME_DA_CONTAGEM = <valor>, usando exatamente os
rótulos POSITIVOS, NEGATIVOS, PARES, IMPARES e INVALIDOS.
'''
invalido = negativo = positivo = par = impar = 0
num = 1
while num != 0:
  try:
    num = int(input())
    if num != 0:
      if num < 0:
        negativo = negativo + 1
      if num > 0:
        positivo = positivo + 1
      if abs(num) % 2 == 0:
        par = par + 1

      if abs(num) % 2 != 0:
        impar = impar + 1
  except:
      invalido = invalido + 1
print(f'POSITOVOS = {positivo}')
print(f'NEGATIVOS = {negativo}')
print(f'PARES = {par}')
print(f'IMPARES = {impar}')
print(f'INVALIDOS = {invalido}')


```
