```python
'''
O conselho de Eltond deseja analisar registros de influência do
Um Anel sobre diferentes viajantes da Terra-média. Cada registro
contém um valor inteiro de 0 a 100, representando o nível de
influência percebido. Valores de 0 a 19 indicam Tranquilo no
Condado, de 20 a 39 indicam Atenção da Sociedade, de 40 a 69
indicam Perigo de Isengard, de 70 a 89 indicam Sombra de Mordor
e de 90 a 100 indicam Olho de Sauron. Ao final, o programa
também deve informar quantos registros foram considerados críticos,
 ou seja, quantos valores foram maiores ou iguais a 70.
 ENTRADA
 A entrada possui diversos valores, umpor linha. A primeira
 linha contém um número inteiro N, indicando quantos registros
 serão avaliados. As próximas N linhas contêm, cada uma, um
 único número inteiro entre 0 e 100, representando o nível de
 influência do Um Anel naquele registro. Não haverá dados
 inválidos na entrada.
 SAIDA
 Para cada registro, o programa deve imprimir uma linha no
 formato Registro <x>: <classificacao>, em que <x> é a posição
 do registro, começando em 1, e <classificacao> é uma das
 menssagens definidas na descrição. Após todos os registros, o
 programa deve imprimir uma linha final no formato Registros
 críticos:<y>, em que<y> é a quantidade de valores maiores ou
 iguais a 70.
 '''
contador = critico = 0
num = int(input())
for _ in range(num):
  registro = int(input())
  if 0 <= registro <= 19:
    nivel = 'Tranquilo no Condado'
  elif 20 <= registro <= 39:
    nivel = 'Atenção da Sociedade'
  elif 40 <= registro <= 69:
    nivel = 'Perigo de Isengard'
  elif 70 <= registro <= 89:
    nivel = 'Sombra de Mordor'
  elif 90 <= registro <= 100:
    nivel = 'Olho de Sauron'
  if registro >= 70:
    critico = critico + 1
  contador = contador + 1
  print(f'Registro {contador}: {nivel}')
print(f'Registros críticos: {critico}')
```
