```python
valor = int(input())

#cedula de R$100
nota_100 = valor // 100
valor = valor % 100

#cedula de 20
nota_20 = valor // 20
valor = valor % 20

#cedula de 5
nota_5 = valor // 5
valor = valor % 5

#cedula de 1
nota_1 = valor

print(f'{nota_100} de 100')
print(f'{nota_20} de 20')
print(f'{nota_5} de 5')
print(f'{nota_1} de 1')
```
